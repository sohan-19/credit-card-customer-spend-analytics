```python
# Select numerical columns for outlier analysis

numeric_cols = [
    'per_capita_income',
    'yearly_income',
    'total_debt',
    'credit_score',
    'amount'
]

# Calculate Q1, Q3 and IQR
Q1 = df[numeric_cols].quantile(0.25)
Q3 = df[numeric_cols].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_summary = pd.DataFrame({
    'Q1': Q1,
    'Q3': Q3,
    'IQR': IQR,
    'Lower_Bound': lower_bound,
    'Upper_Bound': upper_bound
})

outlier_summary
```

```python

outlier_counts = {}

for col in numeric_cols:
    outlier_counts[col] = (
        (df[col] < lower_bound[col]) |
        (df[col] > upper_bound[col])
    ).sum()

outlier_counts = pd.Series(outlier_counts).sort_values(ascending=False)

print(outlier_counts)
```

```python
# What is Recency?
# Recency tells us how recently a customer made a transaction.

customer_data['last_transaction_date'].max()
```

```python
reference_date = customer_data['last_transaction_date'].max()

customer_data['recency'] = (
    reference_date - customer_data['last_transaction_date']
).dt.days
```

```python
customer_data['recency'].describe()
```

```python
print(
    pd.cut(
        customer_data['recency'],
        bins=[-1, 90, 180, 365, 730, float('inf')],
        labels=[
            '0-90 Days',
            '91-180 Days',
            '181-365 Days',
            '366-730 Days',
            '731+ Days'
        ]
    ).value_counts().sort_index()
)
```

```python
print("Recency unique values:", customer_data['recency'].nunique())
print("Frequency unique values:", customer_data['transaction_count'].nunique())
print("Monetary unique values:", customer_data['total_spend'].nunique())
```

```python
customer_data['R_score'] = pd.qcut(
    customer_data['recency'],
    q=5,
    labels=[5, 4, 3, 2, 1],
    duplicates='drop'
)

customer_data['F_score'] = pd.qcut(
    customer_data['transaction_count'].rank(method='first'),
    q=5,
    labels=[1, 2, 3, 4, 5]
)

customer_data['M_score'] = pd.qcut(
    customer_data['total_spend'],
    q=5,
    labels=[1, 2, 3, 4, 5]
)
```
